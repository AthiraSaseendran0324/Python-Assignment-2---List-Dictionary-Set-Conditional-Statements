###List (Creation, Modification and Access):
1. List Creation:

a. Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28]

b. Create a list named name_list with five string elements.

```python
# Create a list with five integer elements
age_list = [21, 22, 18, 25, 24]

# Create a list with five string elements
name_list = ["Athira", "Neeraja", "Nidhin", "Vishnu", "anondita"]

print("Age List:", age_list)
print("Name List:", name_list)
```

2. List Operations / Modifications:

a. Append the string "Yazhini" to name_list.   

b. Insert the element 30 at index 2 in age_list.

c. Remove the string "Yazhini" from name_list.

d. Pop the last element from age_list.

e. Extend the age_list with additional ages [29, 30, 26].

f. Sort age_list in descending order.

g. Find Max age, Min age and sum of all ages from age_list.


```python
# Add Yazhini to the name_list
name_list.append("Yazhini")
print("After append:", name_list)

# Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After insert:", age_list)

# Remove Yazhini from name_list
name_list.remove("Yazhini")
print("After remove:", name_list)

# pop the last element from age_list
age_list.pop()
print("After pop:", age_list)

# Add addiional ages to age_list
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# Sort age_list in descending order
age_list.sort(reverse=True)
print("Descending order:", age_list)

# Find maximum, minimum and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

I used different list methods to modify the lists. append() adds an item at the end, insert() adds an item at a specific index, remove() removes an item by value, pop() removes the last item, and extend() adds multiple items. I used sort(reverse=True) to arrange the ages in descending order.

3. Accessing List Elements:

a. Print the first element of name_list.

b. Print the last element of name_list.

c. Print the elements from index 2 to index 4 in name_list

d. Print the elements of name_list in reverse order.

```python
# Print the first element
print("First element:", name_list[0])

# Print the last element
print("Last element:", name_list[-1])

# Print elements from index 2 to index 4
print("Elements from index 2 to 4:", name_list[2:5])

# Print the list in reverse order
print("Reverse order:", name_list[::-1])
```

###Dictionary (Creation, Modification and Access):

a. Create a dictionary named student_marks that maps the names of five
students to their marks (use scale of from 0 to 100).

b. Access and print the mark of a specific student, of your choice.

c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.

d. Update the mark of any one older student to 82.

e. Use the keys(), values(), and items() methods to print all keys, values, and
key-value pairs in the student_marks dictionary.

```python
# Create a dictionary with five students and their marks
student_marks = {
    "Athira": 85,"Neeraja": 78, "Nidhin": 92, "Vishnu": 88, "Anondita": 75 }

# Access and print the mark of a specific student
print("Athira's mark:", student_marks["Athira"])

# Add a new student
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# Update the mark of any older student
student_marks["Vishnu"] = 82
print("After updating Vishnu's mark:", student_marks)

# Print all keys
print("Keys:", student_marks.keys())

# Print all values
print("Values:", student_marks.values())

# Print all key-value pairs
print("Items:", student_marks.items())
```

I created a dictionary using key-value pairs, where the student's name is the key and the mark is the value. I accessed a student's mark using the key, added a new student, and updated an existing student's mark. I also used keys(), values(), and items() to display the dictionary contents.

###Sets (Operations):
a. Create a set called my_set with following values
['a','e','i','o','u','a','a','i'] .
Analyse the output and provide explanation for the same.

b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide
an explanation.

c. Create two sets:

   set1 with values: {1, 3, 5, 7, 9}

   set2 with values: {2, 3, 5, 8, 10}

d. Compute and print the union and intersection of set1 and set2.

```python
# Create a set with following values
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print("My set:", my_set)
```

A set stores only unique values, so the duplicate values 'a' and 'i' are automatically removed.

```python

my_set[4] = 's'
```

The code gives an error because sets are "unordered" and do not support indexing. Therefore, we cannot access or change a set element using an index such as [4].

```python
# Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Union
print("Union:", set1 | set2)

# Intersection
print("Intersection:", set1 & set2)
```

Union combines all unique elements from both sets. Intersection gives only the elements that are common to both sets.

Performance Category Program:
1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

2. Find the performance category based on the input score using following criteria:

       a. Above Average: Score greater than 7

       b. Average: Score between 4 and 7(both inclusive)

       c. Below Average: Score lesser than 4

3. Output: Print the Performance category

4. Additional Step: You can give a prompt of your choice to each category.
For eg: If score below average “Need

```python
# Get the score from the user
score = int(input("Enter your score (0 to 10): "))

# Check the performance category
if score > 7:
  print("Above Average. Excellent work, Keep it up")
elif score >= 4:
  print("Average. Good effort, Keep practicing")
else:
  print("Below Average. Need to improve your performance. Keep practicing")
```

I used input() to get the score from the user. I used if, elif, and else statements to check the score and determine the performance category. Scores above 7 are Above Average, scores from 4 to 7 are Average, and scores below 4 are Below Average.
